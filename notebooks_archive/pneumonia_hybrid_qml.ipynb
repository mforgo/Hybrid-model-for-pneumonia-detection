## 3. Data
### 3.1 Dataset Description
We use the Chest X-Ray Images (Pneumonia) dataset collected by Mooney *et al.*, available through Kaggle. The dataset comprises 5,856 frontal-view chest radiographs in JPEG format, annotated as either *Normal* or *Pneumonia* by expert radiologists. The original split contains 5,216 training images, 16 validation images, and 624 test images.
### 3.2 Data Repartitioning
The original validation set of 16 images is insufficient for reliable model selection. We merge the original training and validation sets, then perform a stratified 80/20 split to obtain new training and validation partitions. The original test set is preserved without modification to ensure comparability with prior work.
### 3.3 Preprocessing and Augmentation
All images are resized to $224 \times 224$ pixels and normalised using ImageNet statistics. During training, we apply medical-safe augmentations: random rotation ($\pm 7^\circ$), random affine translation ($\pm 5\%$), and colour jitter (brightness and contrast $\pm 20\%$). These augmentations model realistic variations in patient positioning and X-ray exposure parameters. Class imbalance is addressed through square-root weighted sampling.

```python
import kagglehub
DATA_PATH = Path(kagglehub.dataset_download('paultimothymooney/chest-xray-pneumonia'))
def find_dataset_root(base: Path) -> Path:
    for p in sorted(base.rglob('train')):
        if p.is_dir() and '__MACOSX' not in p.parts and (p.parent / 'test').is_dir():
            return p.parent
    raise FileNotFoundError(f'Could not find train/test under {base}')
DATA_PATH = find_dataset_root(DATA_PATH)
expected = {
    ('train', 'NORMAL'): 1341, ('train', 'PNEUMONIA'): 3875,
    ('val',   'NORMAL'):    8, ('val',   'PNEUMONIA'):    8,
    ('test',  'NORMAL'):  234, ('test',  'PNEUMONIA'):  390,
}
for (split, cls), exp in expected.items():
    found = len(list((DATA_PATH / split / cls).glob('*.jpeg')))
    assert found == exp, f'{split}/{cls}: expected {exp}, found {found}'
```

```python
def collect_paths_labels(root: Path, split: str):
    paths, labels = [], []
    for label, cls in enumerate(['NORMAL', 'PNEUMONIA']):
        for p in (root / split / cls).glob('*.jpeg'):
            paths.append(p); labels.append(label)
    return paths, labels
train_p, train_l = collect_paths_labels(DATA_PATH, 'train')
val_p,   val_l   = collect_paths_labels(DATA_PATH, 'val')
test_p,  test_l  = collect_paths_labels(DATA_PATH, 'test')
all_p = train_p + val_p
all_l = train_l + val_l
new_train_p, new_val_p, new_train_l, new_val_l = train_test_split(
    all_p, all_l, test_size=cfg.val_split, stratify=all_l, random_state=SEED
)
```

```python
from PIL import Image
IMAGENET_MEAN = [0.485, 0.456, 0.406]
IMAGENET_STD  = [0.229, 0.224, 0.225]
train_transform = T.Compose([
    T.Resize((cfg.img_size, cfg.img_size)),
    T.RandomRotation(degrees=7),
    T.RandomAffine(degrees=0, translate=(0.05, 0.05)),
    T.ColorJitter(brightness=0.2, contrast=0.2),
    T.ToTensor(),
    T.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])
eval_transform = T.Compose([
    T.Resize((cfg.img_size, cfg.img_size)),
    T.ToTensor(),
    T.Normalize(IMAGENET_MEAN, IMAGENET_STD),
])
class XRayDataset(Dataset):
    def __init__(self, paths, labels, transform):
        self.paths = paths
        self.labels = labels
        self.transform = transform
    def __len__(self):
        return len(self.paths)
    def __getitem__(self, idx):
        img = Image.open(self.paths[idx]).convert('RGB')
        return self.transform(img), self.labels[idx]
train_labels_t = torch.tensor(new_train_l, dtype=torch.long)
class_counts   = torch.bincount(train_labels_t)
class_weights  = 1.0 / torch.sqrt(class_counts.float())
sample_weights = class_weights[train_labels_t]
sampler = WeightedRandomSampler(sample_weights, len(sample_weights), replacement=True)
train_ds = XRayDataset(new_train_p, new_train_l, train_transform)
val_ds   = XRayDataset(new_val_p,   new_val_l,   eval_transform)
test_ds  = XRayDataset(test_p,      test_l,       eval_transform)
train_loader = DataLoader(train_ds, batch_size=cfg.batch_size, sampler=sampler,  num_workers=0, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=64,             shuffle=False,    num_workers=0, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=64,             shuffle=False,    num_workers=0, pin_memory=True)
```

## 3.5 Domain Adaptation: DANN
To mitigate the dataset shift between the training set (74.2% pneumonia) and testing set (62.5% pneumonia), we employ a Domain-Adversarial Neural Network (DANN) with a Gradient Reversal Layer (GRL). This follows the **transductive unsupervised domain adaptation** paradigm (Tzeng *et al.*, 2014; Ganin & Lempitsky, 2015).
### Domain Definitions
- **Source domain (train set):** Labeled chest X-rays with 74.2% pneumonia prevalence. Both images and medical labels are used during training.
- **Target domain (test set):** Unlabeled chest X-rays with 62.5% pneumonia prevalence. **Only images are used**—medical labels are completely hidden from the model during DANN training.
### Why This Is Not Data Leakage
In standard supervised learning, touching the test set is forbidden. However, in transductive UDA, the feature extractor is permitted to observe the *unlabeled* target distribution to learn domain-invariant representations. The domain discriminator receives only binary domain membership (source = 0, target = 1), never the diagnostic Normal/Pneumonia labels. The feature extractor simply adapts to the pixel-level characteristics (contrast, exposure, acquisition style) of the target domain without any knowledge of the ground-truth diagnoses.
The validation set cannot serve as the target domain because it was created via stratified split from the training data, sharing the same 74.2% class imbalance. The domain shift exists exclusively between the train/val cluster and the test cluster.
### The Gradient Reversal Layer
The GRL acts as the identity transform during the forward pass but reverses (and scales by $\alpha$) the gradient during backpropagation:
$$\frac{\partial \mathcal{L}_{\text{domain}}}{\partial \mathbf{x}} = -\alpha \frac{\partial \mathcal{L}_{\text{domain}}}{\partial \mathbf{x}}$$
This forces the feature extractor to produce representations that are indistinguishable between domains, ensuring features encode pathology rather than dataset-specific acquisition bias.

```python
import torch
import torch.nn as nn
from torch.autograd import Function
class GradientReversalLayer(Function):
    @staticmethod
    def forward(ctx, x, alpha):
        ctx.alpha = alpha
        return x.view_as(x)
    @staticmethod
    def backward(ctx, grad_output):
        output = grad_output.neg() * ctx.alpha
        return output, None
class DANN_ConvNeXt(nn.Module):
    def __init__(self):
        super().__init__()
        backbone = torchvision.models.convnext_tiny(weights=ConvNeXt_Tiny_Weights.IMAGENET1K_V1)
        self.features = backbone.features
        self.pool = nn.Sequential(nn.AdaptiveAvgPool2d(1), nn.Flatten())
        self.label_predictor = nn.Sequential(
            nn.Linear(768, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
        self.domain_classifier = nn.Sequential(
            nn.Linear(768, 64),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, 1),
            nn.Sigmoid()
        )
    def forward(self, x, alpha=1.0):
        f = self.pool(self.features(x))
        label_preds = self.label_predictor(f)
        f_reversed = GradientReversalLayer.apply(f, alpha)
        domain_preds = self.domain_classifier(f_reversed)
        return label_preds, domain_preds, f
```

```python
def get_grl_alpha(epoch, total_epochs, schedule='ganin'):
    """
    Compute GRL alpha for a given epoch.
    Schedules:
      'ganin':      2 / (1 + exp(-10*p)) - 1              (original Ganin & Lempitsky)
      'linear_0.5':  0.5 * p                               (linear ramp to 0.5)
      'linear_1.0':  1.0 * p                               (linear ramp to 1.0)
      'step':        0.0 for first 50%, 1.0 after          (step function)
    Args:
        epoch: current epoch (0-indexed)
        total_epochs: total number of epochs
        schedule: string identifier for schedule type
    Returns:
        alpha float
    """
    p = float(epoch) / max(total_epochs - 1, 1)
    if schedule == 'ganin':
        return 2. / (1. + np.exp(-10 * p)) - 1
    elif schedule == 'linear_0.5':
        return 0.5 * p
    elif schedule == 'linear_1.0':
        return 1.0 * p
    elif schedule == 'step':
        return 1.0 if p >= 0.5 else 0.0
    else:
        raise ValueError(f"Unknown GRL schedule: {schedule}")
```

```python
FEAT_DIR = Path(cfg.artifacts_dir) / 'features'
FEAT_DIR.mkdir(parents=True, exist_ok=True)
from tqdm.auto import tqdm
from itertools import cycle
def train_dann_epoch(model, train_loader, test_loader, optimizer, epoch, total_epochs):
    model.train()
    criterion = nn.BCELoss()
    for (source_x, source_y), (target_x, _) in zip(train_loader, cycle(test_loader)):
        source_x, source_y = source_x.to(cfg.device), source_y.to(cfg.device).float()
        torch.cuda.empty_cache()
        target_x = target_x.to(cfg.device)
        p = float(epoch) / total_epochs
        alpha = 2. / (1. + np.exp(-10 * p)) - 1
        optimizer.zero_grad()
        source_domain_y = torch.zeros(source_x.size(0), device=cfg.device)
        src_label_preds, src_domain_preds, _ = model(source_x, alpha)
        loss_label = criterion(src_label_preds.squeeze(), source_y)
        loss_domain_src = criterion(src_domain_preds.squeeze(), source_domain_y)
        target_domain_y = torch.ones(target_x.size(0), device=cfg.device)
        _, tgt_domain_preds, _ = model(target_x, alpha)
        loss_domain_tgt = criterion(tgt_domain_preds.squeeze(), target_domain_y)
        total_loss = loss_label + loss_domain_src + loss_domain_tgt
        total_loss.backward()
        optimizer.step()
DANN_EPOCHS = 10
DANN_CACHE = FEAT_DIR / 'dann_convnext_finetuned.pt'
dann_model = DANN_ConvNeXt().to(cfg.device)
optimizer_dann = torch.optim.Adam(dann_model.parameters(), lr=1e-4)
if DANN_CACHE.exists():
    print('Loading cached DANN-finetuned ConvNeXt...')
    dann_model.load_state_dict(torch.load(DANN_CACHE, weights_only=True))
else:
    print('Training DANN for domain adaptation...')
    print('  Source domain: train set (labels used for classification)')
    print('  Target domain: test set (unlabeled, domain label only)')
    print('  Note: test_loader is cycled to cover all training samples')
    print('  Source domain: train set (labels used for classification)')
    print('  Target domain: test set (unlabeled, domain label only)')
    for epoch in range(DANN_EPOCHS):
        dann_model.train()
        alpha = get_grl_alpha(epoch, DANN_EPOCHS, schedule="ganin")
        running_loss, running_label, running_domain = 0.0, 0.0, 0.0
        n_samples = 0
        for (source_x, source_y), (target_x, _) in zip(train_loader, cycle(test_loader)):
            source_x, source_y = source_x.to(cfg.device), source_y.to(cfg.device).float()
            target_x = target_x.to(cfg.device)
            optimizer_dann.zero_grad()
            source_domain_y = torch.zeros(source_x.size(0), device=cfg.device)
            src_label_preds, src_domain_preds, _ = dann_model(source_x, alpha)
            loss_label = nn.BCELoss()(src_label_preds.squeeze(), source_y)
            loss_domain_src = nn.BCELoss()(src_domain_preds.squeeze(), source_domain_y)
            target_domain_y = torch.ones(target_x.size(0), device=cfg.device)
            _, tgt_domain_preds, _ = dann_model(target_x, alpha)
            loss_domain_tgt = nn.BCELoss()(tgt_domain_preds.squeeze(), target_domain_y)
            total_loss = loss_label + loss_domain_src + loss_domain_tgt
            total_loss.backward()
            optimizer_dann.step()
            bs = source_x.size(0)
            running_loss += total_loss.item() * bs
            running_label += loss_label.item() * bs
            running_domain += (loss_domain_src.item() + loss_domain_tgt.item()) * bs
            n_samples += bs
        avg_loss = running_loss / n_samples
        avg_label = running_label / n_samples
        avg_domain = running_domain / n_samples
        print(f'  Epoch {epoch+1}/{DANN_EPOCHS} (alpha={alpha:.3f}): loss={avg_loss:.4f} label={avg_label:.4f} domain={avg_domain:.4f}')
        p = float(epoch) / DANN_EPOCHS
        alpha = 2. / (1. + np.exp(-10 * p)) - 1
        print(f'  Epoch {epoch+1}/{DANN_EPOCHS} (alpha={alpha:.3f})')
    torch.save(dann_model.state_dict(), DANN_CACHE)
    print('DANN training complete. Model saved.')
```

## 4. Methods
### 4.1 Feature Extraction
We employ a pretrained ConvNeXt-Tiny backbone (Liu *et al.*, 2022) as a frozen feature extractor. The model, originally trained on ImageNet-1K, is modified by replacing the classification head with an adaptive average pooling layer followed by flattening, yielding 768-dimensional feature vectors. The backbone weights are frozen during all subsequent training to prevent overfitting on the relatively small medical dataset.
### 4.2 Dimensionality Reduction
To match the amplitude encoding requirements of the quantum circuit, we reduce the 768-dimensional features to $d = 64$ dimensions using principal component analysis (PCA). The scaler and PCA transformation are fitted exclusively on the training partition to prevent data leakage. The reduced features are subsequently $\ell_2$-normalised to satisfy the unit-norm constraint of quantum state preparation:
$$\|\mathbf{x}\|_2 = 1, \quad \mathbf{x} \in \mathbb{R}^{64}.$$
The choice of 64 dimensions is motivated by the amplitude encoding scheme: $n$ qubits can encode $2^n$ amplitudes, and $2^6 = 64$ provides a balance between information retention and circuit tractability.
### 4.3 Variational Quantum Circuit
#### 4.3.1 Ansatz Architecture
We employ the data re-uploading ansatz introduced by Pérez-Salinas *et al.* (2020), which has been proven to be a universal function approximator. The circuit consists of $L$ layers, each comprising:
1. **Data encoding:** Amplitude embedding of the input vector $\mathbf{x} \in \mathbb{R}^{64}$:
   $$|\mathbf{x}\rangle = \sum_{i=0}^{63} x_i |i\rangle,$$
   where zero-padding is applied if necessary.
2. **Parametrised rotations:** Single-qubit $\mathrm{Rot}(\theta, \phi, \omega)$ gates on each qubit, contributing $3n$ trainable parameters per layer.
3. **Entanglement:** A ring of CNOT gates connecting qubit $i$ to qubit $(i+1) \bmod n$, providing nearest-neighbour entanglement with periodic boundary conditions.
The measurement observable is the Pauli-$Z$ operator on the first qubit, yielding an expectation value in $[-1, 1]$. This local observable choice is motivated by the work of Cerezo *et al.* (2021), who showed that local cost functions mitigate the barren plateau problem.
#### 4.3.2 Loss Function
We use a mean squared error loss with targets mapped to $\{-1, +1\}$:
$$\mathcal{L}(\boldsymbol{\theta}) = \frac{1}{B} \sum_{j=1}^{B} \left(f_{\boldsymbol{\theta}}(\mathbf{x}_j) - y_j\right)^2,$$
where $f_{\boldsymbol{\theta}}(\mathbf{x}) \in [-1, 1]$ is the circuit output and $y_j \in \{-1, +1\}$ is the encoded label. Probabilities for evaluation are obtained via the affine transformation $p = (f + 1)/2$.
#### 4.3.3 Optimisation
Gradients are computed using the adjoint differentiation method, which provides exact gradients with a single additional circuit evaluation per parameter. Optimisation is performed with Adam ($\beta_1 = 0.9$, $\beta_2 = 0.999$, $\epsilon = 10^{-8}$) and a cosine annealing learning rate schedule with linear warmup over the first 3 epochs.

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
import joblib
import hashlib
class SupervisedVAE(nn.Module):
    """Variational Autoencoder with auxiliary classification head.
    
    Extends the standard VAE with a classifier on the latent space:
        L_total = L_recon + beta * L_KL + lambda_clf * L_clf
    
    The classification head ensures the latent space preserves
    class-discriminative features during dimensionality reduction.
    
    Architecture:
        Encoder: Linear(768 -> 256) -> LeakyReLU -> BatchNorm -> Dropout
                 -> fc_mu(256 -> 64), fc_logvar(256 -> 64)
        Decoder: Linear(64 -> 256) -> BatchNorm -> LeakyReLU -> Dropout -> Linear(256 -> 768)
        Classifier: Linear(64 -> 32) -> ReLU -> Dropout(0.2) -> Linear(32 -> 1) -> Sigmoid
    """
    def __init__(self, input_dim=768, latent_dim=64, beta=0.001):
        super().__init__()
        self.beta = beta
        
        # Encoder: 768 -> 256 -> latent_dim (outputs mu and logvar)
        self.fc1 = nn.Linear(input_dim, 256)
        self.bn1 = nn.BatchNorm1d(256)
        self.drop1 = nn.Dropout(0.2)
        self.fc_mu = nn.Linear(256, latent_dim)
        self.fc_logvar = nn.Linear(256, latent_dim)
        
        # Decoder: latent_dim -> 256 -> 768
        self.decoder = nn.Sequential(
            nn.Linear(latent_dim, 256),
            nn.BatchNorm1d(256),
            nn.LeakyReLU(0.2),
            nn.Dropout(0.2),
            nn.Linear(256, input_dim)
        )
        
        # Classification head on latent space
        self.classifier = nn.Sequential(
            nn.Linear(latent_dim, 32),
            nn.ReLU(),
            nn.Dropout(0.2),
            nn.Linear(32, 1),
            nn.Sigmoid()
        )
    
    def encode(self, x):
        """Encode input to latent space using mean (deterministic for inference).
        
        Backward-compatible with old FeatureAutoencoder.encode()
        and FeatureVAE.encode()
        """
        h = self.drop1(torch.nn.functional.leaky_relu(self.bn1(self.fc1(x)), 0.2))
        return self.fc_mu(h)
    
    def reparameterize(self, mu, logvar):
        """Reparameterization trick: z = mu + eps * std where eps ~ N(0,I)."""
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std
    
    def kl_loss(self, mu, logvar):
        """KL divergence: -0.5 * sum(1 + logvar - mu^2 - exp(logvar))"""
        return -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())
    
    def classify(self, z):
        """Classify from latent representation. Returns probability in [0,1]."""
        return self.classifier(z).squeeze(-1)
    
    def forward(self, x):
        """VAE forward pass returning dict with all outputs.
        
        Returns dict with keys:
            z: latent sample (or mu during inference)
            reconstructed: decoder output
            mu, logvar: VAE parameters
            kl: KL divergence loss
            logits: classification logits (sigmoid output)
        """
        h = self.drop1(torch.nn.functional.leaky_relu(self.bn1(self.fc1(x)), 0.2))
        mu = self.fc_mu(h)
        logvar = self.fc_logvar(h)
        z = self.reparameterize(mu, logvar)
        reconstructed = self.decoder(z)
        kl = self.kl_loss(mu, logvar)
        logits = self.classifier(z).squeeze(-1)
        
        return {
            'z': z,
            'reconstructed': reconstructed,
            'mu': mu,
            'logvar': logvar,
            'kl': kl,
            'logits': logits
        }
```

```python
import torch
def coral_loss(feats_src, feats_tgt):
    """
    CORrelation ALignment loss.
    Minimizes Frobenius norm between source and target covariance matrices.
    L_coral = (1 / 4 * d^2) * ||Cov(X_src) - Cov(X_tgt)||²_F
    Args:
        feats_src: (batch, d) source features
        feats_tgt: (batch, d) target features
    Returns:
        scalar loss
    """
    d = feats_src.size(-1)
    # Center
    src = feats_src - feats_src.mean(0, keepdim=True)
    tgt = feats_tgt - feats_tgt.mean(0, keepdim=True)
    # Covariance matrices
    cov_src = (src.T @ src) / (src.size(0) - 1)
    cov_tgt = (tgt.T @ tgt) / (tgt.size(0) - 1)
    # Frobenius norm squared, normalized
    loss = (cov_src - cov_tgt).pow(2).sum() / (4 * d * d)
    return loss
def coral_loss_strided(feats_src_chunks, feats_tgt_chunks):
    """
    CORAL loss aggregated over multiple source/target chunk pairs.
    Useful for minibatch CORAL where full covariance estimation
    is done over accumulated features across multiple batches.
    """
    src_all = torch.cat(feats_src_chunks, dim=0)
    tgt_all = torch.cat(feats_tgt_chunks, dim=0)
    return coral_loss(src_all, tgt_all)
```

```python
def fit_vae_pipeline(X_train, X_val, X_test, latent_dim, save_dir, prefix,
                     device='cuda', epochs=30, batch_size=32, lr=1e-3, beta=0.001,
                     lambda_coral=0.0, lambda_clf=0.0,
                     y_train=None, y_val=None, y_test_labels=None, pos_weight=None):
    """Train VAE and return L2-normalized latent features.
    Uses VAE loss: recon_loss + beta * kl_loss
    Optionally adds CORAL loss for domain alignment: total_loss += lambda_coral * coral_loss(z_src, z_tgt)
    Optionally adds supervised classification loss: total_loss += lambda_clf * BCE_classification
    Contract matches the old fit_ae_pipeline():
        X_train, X_val, X_test (np.ndarray, shape (N, 768))
        y_train, y_val, y_test_labels: labels (required if lambda_clf > 0)
        Returns: latent_train, latent_val, latent_test, scaler, vae_model, losses
    Args:
        lambda_coral: Weight of CORAL domain alignment loss (0 = disabled).
                      When > 0, aligns latent distributions of training (source)
                      and test (target) sets via CORAL loss.
        lambda_clf: Weight of supervised classification loss (0 = unsupervised VAE).
                    When > 0, y_train and y_val must be provided.
        y_train, y_val: Training/validation labels (0/1). Required if lambda_clf > 0.
        y_test_labels: Test labels (for cache validation metadata).
        pos_weight: Tensor for BCEWithLogitsLoss pos_weight.
    """
    # supervised_vae_loss is defined in notebook global scope (cell 22)
    save_dir = Path(save_dir)
    
    # Auto-adjust prefix for supervised variant
    eff_prefix = f'{prefix}_svae' if lambda_clf > 0 else prefix
    
    cache_train = save_dir / f'{eff_prefix}_vae_train.npy'
    cache_meta  = save_dir / f'{eff_prefix}_vae_meta.json'
    
    feat_hash = hashlib.md5(X_train[:100].tobytes()).hexdigest()[:8]
    
    if cache_train.exists() and cache_meta.exists():
        import json as _json
        with open(cache_meta) as f:
            meta = _json.load(f)
        if (meta.get('feat_hash') == feat_hash
            and meta.get('latent_dim') == latent_dim
            and meta.get('lambda_coral') == lambda_coral
            and meta.get('lambda_clf') == lambda_clf):
            print(f'Loading cached VAE features (hash={feat_hash})...')
            Xtr = np.load(save_dir / f'{eff_prefix}_vae_train.npy')
            Xva = np.load(save_dir / f'{eff_prefix}_vae_val.npy')
            Xte = np.load(save_dir / f'{eff_prefix}_vae_test.npy')
            scaler = joblib.load(save_dir / f'{eff_prefix}_vae_scaler.pkl')
            vae = SupervisedVAE(input_dim=X_train.shape[1], latent_dim=latent_dim, beta=beta)
            w_path = save_dir / f'{eff_prefix}_vae_weights.pt'
            if w_path.exists():
                vae.load_state_dict(torch.load(w_path, map_location='cpu'))
            losses = meta.get('losses', [])
            print(f'  VAE loss (cached): {meta.get("final_loss", "N/A")}')
            return Xtr, Xva, Xte, scaler, vae, losses
        else:
            print(f'Cache stale. Retraining VAE...')
    else:
        print('No VAE cache found. Training...')
    
    # Standardize input features
    scaler = StandardScaler()
    Xtr_scaled = scaler.fit_transform(X_train)
    Xva_scaled = scaler.transform(X_val)
    Xte_scaled = scaler.transform(X_test)
    
    # Convert to tensors
    train_tensor = torch.tensor(Xtr_scaled, dtype=torch.float32).to(device)
    val_tensor   = torch.tensor(Xva_scaled, dtype=torch.float32).to(device)
    test_tensor  = torch.tensor(Xte_scaled, dtype=torch.float32).to(device)
    
    # Build DataLoader — supervised or unsupervised
    is_supervised = lambda_clf > 0 and y_train is not None
    if is_supervised:
        y_train_t = torch.tensor(y_train, dtype=torch.float32).to(device)
        train_dataset = TensorDataset(train_tensor, y_train_t)
    else:
        train_dataset = TensorDataset(train_tensor, train_tensor)
    ae_train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
    
    # Initialize VAE
    vae = SupervisedVAE(input_dim=X_train.shape[1], latent_dim=latent_dim, beta=beta).to(device)
    optimizer_vae = optim.Adam(vae.parameters(), lr=lr, weight_decay=1e-5)
    criterion_vae = nn.MSELoss()
    
    # Training loop
    mode_str = 'SupervisedVAE' if is_supervised else 'VAE'
    print(f"Training {mode_str} ({X_train.shape[1]} -> {latent_dim}, beta={beta}) for {epochs} epochs...")
    if lambda_coral > 0:
        print(f"  CORAL domain alignment enabled (lambda_coral={lambda_coral})")
    if is_supervised:
        print(f"  Classification loss enabled (lambda_clf={lambda_clf})")
    vae.train()
    final_loss = None
    train_losses = []
    
    for epoch in range(epochs):
        epoch_loss = 0.0
        epoch_recon = 0.0
        epoch_kl = 0.0
        epoch_coral = 0.0
        epoch_clf = 0.0
        
        for batch in ae_train_loader:
            if is_supervised:
                batch_x, batch_y = batch
            else:
                batch_x = batch[0]
                batch_y = None
            
            optimizer_vae.zero_grad()
            output = vae(batch_x)
            z = output['z']
            reconstructed = output['reconstructed']
            kl = output['kl']
            
            if is_supervised:
                total_loss, loss_comp = supervised_vae_loss(
                    output, batch_x, batch_y,
                    beta=beta, lambda_clf=lambda_clf, pos_weight=pos_weight
                )
                recon_loss_val = loss_comp['recon']
                kl_loss_val = loss_comp['kl']
                clf_loss_val = loss_comp['clf']
            else:
                recon_loss = criterion_vae(reconstructed, batch_x)
                kl_loss_val = kl
                total_loss = recon_loss + beta * kl
                recon_loss_val = recon_loss.item()
                clf_loss_val = 0.0
            
            # CORAL domain alignment on latent space
            coral_loss_val = 0.0
            if lambda_coral > 0 and X_test is not None:
                test_batch_size = batch_x.size(0)
                test_idx = np.random.choice(len(X_test), size=test_batch_size, replace=False)
                test_sample = torch.tensor(X_test[test_idx], dtype=torch.float32).to(device)
                
                with torch.no_grad():
                    test_z = vae.encode(test_sample)
                
                coral_loss_val = lambda_coral * coral_loss(z, test_z)
                total_loss = total_loss + coral_loss_val
            
            total_loss.backward()
            optimizer_vae.step()
            
            epoch_loss += total_loss.item()
            epoch_recon += recon_loss_val
            epoch_kl += kl_loss_val.item() if hasattr(kl_loss_val, 'item') else kl_loss_val
            epoch_coral += coral_loss_val if isinstance(coral_loss_val, (int, float)) else coral_loss_val.item()
            epoch_clf += clf_loss_val
        
        n_batches = len(ae_train_loader)
        avg_loss = epoch_loss / n_batches
        avg_recon = epoch_recon / n_batches
        avg_kl = epoch_kl / n_batches
        avg_coral = epoch_coral / n_batches
        avg_clf = epoch_clf / n_batches
        train_losses.append(avg_loss)
        
        if (epoch + 1) % 10 == 0:
            parts = [f"Total: {avg_loss:.4f}", f"Recon: {avg_recon:.4f}", f"KL: {avg_kl:.6f}"]
            if avg_coral > 0:
                parts.append(f"Coral: {avg_coral:.6f}")
            if avg_clf > 0:
                parts.append(f"Clf: {avg_clf:.6f}")
            print(f"  {mode_str} Epoch {epoch+1}/{epochs} | {' | '.join(parts)}")
        final_loss = avg_loss
    
    # Extract latent features using mean (deterministic for inference)
    vae.eval()
    with torch.no_grad():
        latent_train = vae.encode(train_tensor).cpu().numpy()
        latent_val   = vae.encode(val_tensor).cpu().numpy()
        latent_test  = vae.encode(test_tensor).cpu().numpy()
    
    # L2 normalize for quantum amplitude encoding
    latent_train = normalize(latent_train, norm='l2')
    latent_val   = normalize(latent_val, norm='l2')
    latent_test  = normalize(latent_test, norm='l2')
    
    # Save cache
    np.save(save_dir / f'{eff_prefix}_vae_train.npy', latent_train)
    np.save(save_dir / f'{eff_prefix}_vae_val.npy',   latent_val)
    np.save(save_dir / f'{eff_prefix}_vae_test.npy',  latent_test)
    joblib.dump(scaler, save_dir / f'{eff_prefix}_vae_scaler.pkl')
    torch.save(vae.state_dict(), save_dir / f'{eff_prefix}_vae_weights.pt')
    
    import json as _json
    with open(cache_meta, 'w') as f:
        _json.dump({
            'feat_hash': feat_hash,
            'latent_dim': latent_dim,
            'beta': beta,
            'lambda_coral': lambda_coral,
            'lambda_clf': lambda_clf,
            'final_loss': float(final_loss),
            'epochs': epochs,
            'losses': train_losses
        }, f)
    
    print(f'  Final {mode_str} total loss: {final_loss:.4f}')
    print(f'  Latent feature shapes: train={latent_train.shape}, val={latent_val.shape}, test={latent_test.shape}')
    return latent_train, latent_val, latent_test, scaler, vae, train_losses
```

```python
def supervised_vae_loss(output, target_x, target_y, beta=0.001, lambda_clf=0.0, pos_weight=None):
    """
    Combined loss for SupervisedVAE.
    
    L_total = MSE_recon + beta * KL + lambda_clf * BCE_classification
    
    Args:
        output: dict from SupervisedVAE.forward()
        target_x: original input features (for reconstruction)
        target_y: class labels (0 or 1)
        beta: KL loss weight
        lambda_clf: classification loss weight (0 = unsupervised VAE)
        pos_weight: tensor for BCEWithLogitsLoss pos_weight (handles imbalance)
    Returns:
        total_loss: scalar
        losses: dict with 'recon', 'kl', 'clf' components
    """
    recon_loss = nn.MSELoss()(output['reconstructed'], target_x)
    kl_loss_val = output['kl']
    
    total = recon_loss + beta * kl_loss_val
    losses = {'recon': recon_loss.item(), 'kl': kl_loss_val.item(), 'clf': 0.0}
    
    if lambda_clf > 0 and target_y is not None:
        if pos_weight is not None:
            clf_criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)
        else:
            clf_criterion = nn.BCELoss()
        clf_loss = clf_criterion(output['logits'], target_y)
        total = total + lambda_clf * clf_loss
        losses['clf'] = clf_loss.item()
    
    return total, losses
def compute_pos_weight(y_train):
    """Compute BCEWithLogitsLoss pos_weight from training labels.
    
    pos_weight = (n_negative / n_positive)  -- handles class imbalance
    """
    n_pos = (y_train == 1).sum()
    n_neg = (y_train == 0).sum()
    return torch.tensor(n_neg / n_pos, dtype=torch.float32)
```

```python
import numpy as np
def inter_domain_mixup(feats_src, feats_tgt, labels_src, alpha=0.2):
    """
    Feature-space MixUp between source and target domains.
    
    Convex combination: x_mixed = λ * src + (1-λ) * tgt
    Labels stay as source labels (transductive UDA setting).
    
    Args:
        feats_src: (batch, d) source features
        feats_tgt: (batch, d) target features  
        labels_src: (batch,) source labels
        alpha: Beta distribution parameter (default 0.2)
    Returns:
        mixed_feats: (batch, d) mixed features
        labels_src: (batch,) unchanged source labels
        lam: (batch,) mixing coefficients per sample
    """
    batch_size = feats_src.shape[0]
    
    # Sample lambda from Beta(alpha, alpha)
    lam = np.random.beta(alpha, alpha, size=batch_size)
    lam = np.maximum(lam, 1.0 - lam)  # bias toward source (optional)
    lam = lam.reshape(-1, 1)
    
    # Random permutation of target features
    perm = np.random.permutation(batch_size)
    feats_tgt_perm = feats_tgt[perm]
    
    # Mix
    mixed = lam * feats_src + (1.0 - lam) * feats_tgt_perm
    
    return mixed, labels_src, lam.squeeze(-1)
def manifold_mixup(feats_src, feats_tgt, labels_src, alpha=0.2):
    """
    Manifold MixUp: interpolate in feature space with per-sample λ.
    
    Unlike standard MixUp, λ is sampled per sample (not shared across batch),
    and mixing happens at the feature level rather than input level.
    
    Args:
        feats_src: (batch, d) source features
        feats_tgt: (batch, d) target features
        labels_src: (batch,) source labels
        alpha: Beta distribution parameter
    Returns:
        mixed_feats: (batch, d) mixed features
        labels_src: (batch,) unchanged labels  
        lam: (batch,) mixing coefficients
    """
    batch_size = feats_src.shape[0]
    
    # Per-sample lambda
    lam = np.random.beta(alpha, alpha, size=batch_size)
    lam = lam.reshape(-1, 1)
    
    # Permute target
    perm = np.random.permutation(batch_size)
    feats_tgt_perm = feats_tgt[perm]
    
    # Manifold mix
    mixed = lam * feats_src + (1.0 - lam) * feats_tgt_perm
    
    return mixed, labels_src, lam.squeeze(-1)
```

## 5. Ansatz Analysis
The performance of a variational quantum circuit depends critically on its expressibility—the ability to approximate arbitrary unitary transformations—and its entanglement capability. Following Sim *et al.* (2019), we quantify these properties before committing to a circuit depth.
### 5.1 Expressibility
The expressibility of an ansatz $\mathcal{A}$ is measured by the Kullback–Leibler divergence between the distribution of state fidelities generated by random parameter settings and the Haar-random distribution:
$$\mathrm{Expr}(\mathcal{A}) = D_{\mathrm{KL}}(P_{\mathrm{ansatz}} \| P_{\mathrm{Haar}}),$$
where $P_{\mathrm{Haar}}$ follows a $\mathrm{Beta}(1, 2^n - 1)$ distribution for $n$ qubits. Lower values indicate greater expressibility.
### 5.2 Entanglement Capability
The entanglement capability is quantified using the Meyer–Wallach measure:
$$\mathrm{Ent}(\mathcal{A}) = \frac{2}{n} \sum_{k=1}^{n} \left(1 - \mathrm{Tr}(\rho_k^2)\right),$$
where $\rho_k$ is the reduced density matrix of qubit $k$. This measures the average entanglement generated across all single-qubit bipartitions.
We sweep $L \in \{1, 2, 3, 4\}$ layers and select the depth that provides the best trade-off between expressibility and parameter count.

```python
from scipy.stats import beta as beta_dist
from scipy.special import kl_div
N_QUBITS = cfg.n_qubits
N_BINS = 75
N_SAMPLES = 2000
def make_vqc_circuit(n_layers, n_qubits=N_QUBITS):
    dev = qml.device('default.qubit', wires=n_qubits)
    @qml.qnode(dev)
    def circuit(x, params):
        qml.AmplitudeEmbedding(x, wires=range(n_qubits), pad_with=0.0)
        for l in range(n_layers):
            for w in range(n_qubits):
                qml.RY(x[w] * np.pi, wires=w)
            for w in range(n_qubits):
                qml.Rot(params[l, w, 0], params[l, w, 1], params[l, w, 2], wires=w)
            for w in range(n_qubits):
                qml.CNOT(wires=[w, (w + 1) % n_qubits])
        return qml.state()
    return circuit
def compute_expressibility(circuit, n_layers, n_samples=N_SAMPLES, n_bins=N_BINS):
    """Compute expressibility as KL divergence from Haar distribution."""
    n_qubits = N_QUBITS
    dim = 2 ** n_qubits
    n_params_per_layer = n_qubits * 3
    total_params = n_layers * n_params_per_layer
    fidelities = []
    for _ in range(n_samples):
        params_a = np.random.uniform(0, 2 * np.pi, (n_layers, n_qubits, 3))
        params_b = np.random.uniform(0, 2 * np.pi, (n_layers, n_qubits, 3))
        x = np.random.randn(dim)
        x = x / np.linalg.norm(x)
        state_a = circuit(x, params_a)
        state_b = circuit(x, params_b)
        fid = np.abs(np.vdot(state_a, state_b)) ** 2
        fidelities.append(fid)
    fidelities = np.array(fidelities)
    # Haar distribution: Beta(1, dim - 1)
    hist_ansatz, bin_edges = np.histogram(fidelities, bins=n_bins, range=(0, 1), density=True)
    bin_centers = (bin_edges[:-1] + bin_edges[1:]) / 2
    p_ansatz = hist_ansatz + 1e-12
    p_ansatz /= p_ansatz.sum()
    p_haar = beta_dist.pdf(bin_centers, 1, dim - 1)
    p_haar = p_haar + 1e-12
    p_haar /= p_haar.sum()
    kl = np.sum(kl_div(p_ansatz, p_haar))
    return kl
def compute_entanglement_capability(circuit, n_layers, n_samples=N_SAMPLES):
    """Compute entanglement capability using Meyer-Wallach measure."""
    n_qubits = N_QUBITS
    dim = 2 ** n_qubits
    mw_measures = []
    for _ in range(n_samples):
        params = np.random.uniform(0, 2 * np.pi, (n_layers, n_qubits, 3))
        x = np.random.randn(dim)
        x = x / np.linalg.norm(x)
        state = circuit(x, params)
        rho = np.outer(state, np.conj(state))
        q = 0.0
        for k in range(n_qubits):
            # Trace out all qubits except qubit k
            remaining = list(range(n_qubits))
            remaining.remove(k)
            # Reshape to tensor, trace over remaining qubits
            rho_tensor = rho.reshape([2] * n_qubits * 2)
            # Move k-th qubit indices to front for both bra and ket
            axes_bra = [k] + remaining
            axes_ket = [k + n_qubits] + [i + n_qubits for i in remaining]
            rho_tensor = np.transpose(rho_tensor, axes_bra + axes_ket)
            rho_tensor = rho_tensor.reshape([2, 2**(n_qubits-1), 2, 2**(n_qubits-1)])
            rho_k = np.trace(rho_tensor, axis1=1, axis2=3)
            purity = np.real(np.trace(rho_k @ rho_k))
            q += 1.0 - purity
        q *= 2.0 / n_qubits
        mw_measures.append(q)
    return np.mean(mw_measures)
```

```python
sweep_results = []
for n_l in tqdm([1, 2, 3, 4], desc='Ansatz analysis sweep'):
    circ = make_vqc_circuit(n_l)
    expr = compute_expressibility(circ, n_l)
    ent  = compute_entanglement_capability(circ, n_l)
    sweep_results.append({'n_layers': n_l, 'Expr(A)': round(expr, 4), 'Ent(A)': round(ent, 4),
                          'n_params': n_l * N_QUBITS * 3})
df_sweep = pd.DataFrame(sweep_results)
df_sweep.to_csv(f'{cfg.results_dir}/expressibility_sweep.csv', index=False)
```

## 6. Classical Baseline: MLP
As a classical comparator, we train a small multilayer perceptron on the same 64-dimensional PCA features. The architecture is $64 \to 32 \to 1$ with ReLU activation, dropout ($p = 0.3$), and sigmoid output. This model serves as the primary baseline against which the VQC is compared.

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(64, 32), nn.ReLU(), nn.Dropout(0.3),
            nn.Linear(32, 1), nn.Sigmoid()
        )
    def forward(self, x):
        return self.net(x).squeeze(-1)
mlp = MLP().to(cfg.device)
n_params_mlp = sum(p.numel() for p in mlp.parameters())
```

```python
from torch.utils.data import TensorDataset
def make_tensor_loaders(X_tr, y_tr, X_va, y_va, batch_size=cfg.batch_size):
    tr_ds = TensorDataset(torch.tensor(X_tr, dtype=torch.float32),
                          torch.tensor(y_tr, dtype=torch.float32))
    va_ds = TensorDataset(torch.tensor(X_va, dtype=torch.float32),
                          torch.tensor(y_va, dtype=torch.float32))
    return (DataLoader(tr_ds, batch_size=batch_size, shuffle=True),
            DataLoader(va_ds, batch_size=256, shuffle=False))
mlp_train_loader, mlp_val_loader = make_tensor_loaders(X_train, y_train, X_val, y_val)
optimizer_mlp = torch.optim.Adam(mlp.parameters(), lr=cfg.learning_rate)
scheduler_mlp = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer_mlp, T_max=cfg.epochs, eta_min=cfg.lr_min
)
criterion_mlp = nn.BCELoss()
def train_mlp(model, train_ld, val_ld, optimizer, scheduler, criterion,
              epochs=cfg.epochs, patience=cfg.early_stopping_patience, device=cfg.device,
              X_test=None, lambda_coral=0.0, mixup_alpha=0.0):
    history = {'train_loss': [], 'val_loss': [], 'val_auc': [], 'coral_loss': []}
    best_val_loss, best_state, wait = float('inf'), None, 0
    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        for xb, yb in train_ld:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * len(xb)
            # MixUp: augment with mixed train-test features
            if mixup_alpha > 0 and X_test is not None:
                test_idx = np.random.choice(len(X_test), size=xb.size(0), replace=False)
                X_test_batch = torch.tensor(X_test[test_idx], dtype=torch.float32).to(device)
                X_mixed, _, _ = inter_domain_mixup(
                    xb.detach().cpu().numpy(),
                    X_test_batch.detach().cpu().numpy(),
                    yb.detach().cpu().numpy(),
                    alpha=mixup_alpha
                )
                X_mixed_t = torch.tensor(X_mixed, dtype=torch.float32).to(device)
                optimizer.zero_grad()
                loss_mixed = criterion(model(X_mixed_t), yb)
                loss_mixed.backward()
                optimizer.step()
                running_loss += loss_mixed.item() * len(xb)
            # CORAL on hidden features (monitoring only)
            if lambda_coral > 0 and X_test is not None:
                with torch.no_grad():
                    train_hidden = model.net[0](xb)
                    train_hidden = torch.relu(train_hidden)
                    test_idx = np.random.choice(len(X_test), size=xb.size(0), replace=False)
                    test_t = torch.tensor(X_test[test_idx], dtype=torch.float32).to(device)
                    test_hidden = model.net[0](test_t)
                    test_hidden = torch.relu(test_hidden)
                coral_val = lambda_coral * coral_loss(train_hidden, test_hidden)
                history['coral_loss'].append(float(coral_val))
        scheduler.step()
        model.eval()
        val_loss, all_probs, all_labels = 0.0, [], []
        with torch.no_grad():
            for xb, yb in val_ld:
                xb, yb = xb.to(device), yb.to(device)
                probs = model(xb)
                val_loss += criterion(probs, yb).item() * len(xb)
                all_probs.extend(probs.cpu().numpy())
                all_labels.extend(yb.cpu().numpy())
        train_l = running_loss / len(train_ld.dataset)
        val_l   = val_loss / len(val_ld.dataset)
        val_auc = roc_auc_score(all_labels, all_probs)
        history['train_loss'].append(train_l)
        history['val_loss'].append(val_l)
        history['val_auc'].append(val_auc)
        if val_l < best_val_loss:
            best_val_loss, best_state, wait = val_l, {k: v.clone() for k, v in model.state_dict().items()}, 0
        else:
            wait += 1
            if wait >= patience:
                break
    model.load_state_dict(best_state)
    return history
mlp_history = train_mlp(mlp, mlp_train_loader, mlp_val_loader,
                        optimizer_mlp, scheduler_mlp, criterion_mlp,
                        X_test=X_test, lambda_coral=0.1, mixup_alpha=0.2)
```

## 7. Quantum Classifier: VQC Training
The VQC uses $n = 6$ qubits with $L = 3$ layers, yielding 108 trainable parameters. The adjoint differentiation method is used for gradient computation. The PennyLane backend is selected automatically, preferring `lightning.gpu` when a CUDA-capable GPU is available.

```python
# --- PennyLane device selection (mode-aware) ---
# Returns:
#   SIM mode: lightning.gpu / lightning.qubit / default.qubit (ideal simulator)
#   QPU/BOTH modes: also initializes a remote IBM device handle for inference
def get_pennylane_device(n_qubits):
    for dev_name in ['lightning.gpu', 'lightning.qubit', 'default.qubit']:
        try:
            dev = qml.device(dev_name, wires=n_qubits)
            print(f'Using PennyLane device: {dev_name}')
            if dev_name == 'default.qubit':
                print('WARNING: Falling back to default.qubit (CPU). VQC training will be VERY slow.')
            return dev
        except Exception as e:
            print(f'  {dev_name} failed: {e}')
            continue
    raise RuntimeError('No suitable PennyLane device found.')
dev_vqc = get_pennylane_device(cfg.n_qubits)
print(f'Run mode: {cfg.run_mode.value}')
# For QPU modes: set up IBM Quantum backend reference (inference only)
qpu_backend = None
if cfg.run_mode in (RunMode.QPU, RunMode.BOTH):
    try:
        from qiskit_ibm_runtime import QiskitRuntimeService
        service = QiskitRuntimeService(
            token=cfg.ibm_token,
            instance=cfg.ibm_instance
        )
        if cfg.ibm_backend:
            qpu_backend = service.backend(cfg.ibm_backend)
        else:
            # 2026: brisbane/sherbrooke retired; auto-select least-busy Heron (e.g. ibm_kingston)
            qpu_backend = service.least_busy(operational=True, simulator=False, min_num_qubits=6)
        print(f'QPU backend: {qpu_backend.name} ({qpu_backend.status().status_msg})')
    except Exception as e:
        print(f'WARNING: Could not connect to IBM Quantum: {e}')
        print('  Falling back to SIM-only mode.')
        cfg.run_mode = RunMode.SIM
```

```python
N_LAYERS = cfg.n_layers
N_QUBITS = cfg.n_qubits
@qml.qnode(dev_vqc, diff_method='adjoint')
def vqc_circuit(x, params):
    # Encode input ONCE via amplitude embedding
    qml.AmplitudeEmbedding(x, wires=range(N_QUBITS), pad_with=0.0)  # input already L2-normalized by PCA pipeline
    # Data re-uploading layers: angle encoding + trainable rotations + entanglement
    for l in range(N_LAYERS):
        # Re-upload input data via RY rotations (angle encoding)
        for w in range(N_QUBITS):
            qml.RY(x[w] * np.pi, wires=w)
        # Trainable single-qubit rotations
        for w in range(N_QUBITS):
            qml.Rot(params[l][w][0], params[l][w][1], params[l][w][2], wires=w)
        # Entangling CNOT ring
        for w in range(N_QUBITS):
            qml.CNOT(wires=[w, (w + 1) % N_QUBITS])
    return qml.expval(qml.PauliZ(0))
n_params_vqc = N_LAYERS * N_QUBITS * 3
print(f'VQC circuit (data re-uploading): {N_LAYERS} layers, {N_QUBITS} qubits, {n_params_vqc} trainable params')
```

```python
def batch_loss(params, X_batch, y_batch):
    """Compute averaged MSE loss over a batch."""
    total_loss = 0.0
    for x, y in zip(X_batch, y_batch):
        raw = vqc_circuit(x, params)  # NO float() — breaks autograd
        target = 1.0 if y == 1 else -1.0
        total_loss += (raw - target) ** 2
    return total_loss / len(X_batch)
# Create the gradient function ONCE
batch_grad_fn = qml.grad(batch_loss, argnum=0)
def forward_and_grad(params, X_batch, y_batch):
    loss = batch_loss(params, X_batch, y_batch)
    grad = batch_grad_fn(params, X_batch, y_batch)
    return loss, grad
def lr_cosine_warmup(epoch, lr0=cfg.learning_rate, lr_min=cfg.lr_min,
                     warmup=cfg.warmup_epochs, total=cfg.epochs):
    if epoch < warmup:
        return lr0 * (epoch + 1) / warmup
    t = (epoch - warmup) / max(total - warmup, 1)
    return lr_min + 0.5 * (lr0 - lr_min) * (1 + np.cos(np.pi * t))
```

```python
import time
PARAMS_SAVE = Path(cfg.results_dir) / 'vqc_best_params.npy'
if PARAMS_SAVE.exists():
    print('Loading cached VQC parameters...')
    best_params = np.load(PARAMS_SAVE)
    print('  WARNING: Training history not available for cached params.')
    print('  Delete vqc_best_params.npy to retrain from scratch.')
else:
    np.random.seed(SEED)
    params = np.random.uniform(0, 2*np.pi, (N_LAYERS, N_QUBITS, 3))
    m  = np.zeros_like(params)
    v  = np.zeros_like(params)
    b1, b2, eps_adam = 0.9, 0.999, 1e-8
    t  = 0
    counts = np.bincount(y_train.astype(int))
    class_weights = 1.0 / np.sqrt(counts)
    sample_weights = class_weights[y_train.astype(int)]
    p_weights = sample_weights / np.sum(sample_weights)
    history_vqc = {'train_loss': [], 'val_auc': [], 'val_loss': [], 'coral_loss': []}
    best_val_auc, best_params, wait = -1.0, params.copy(), 0
    total_batches = int(np.ceil(len(X_train) / cfg.batch_size))
    # Domain adaptation: CORAL + MixUp
    LAMBDA_CORAL = 0.1       # CORAL loss weight (0 = disable)
    MIXUP_ALPHA = 0.2        # MixUp Beta parameter (0 = disable)
    print(f'Starting VQC training: {cfg.epochs} epochs, {total_batches} batches/epoch')
    print(f'  LR={cfg.learning_rate}, warmup={cfg.warmup_epochs}, patience={cfg.early_stopping_patience}')
    print(f'  Train: {len(X_train)}, Val: {len(X_val)}')
    print('-' * 85)
    print(f'{"Epoch":>5} | {"LR":>8} | {"Train Loss":>12} | {"Val AUC":>8} | {"Wait":>4} | {"Time":>8}')
    print('-' * 85)
    for epoch in range(cfg.epochs):
        t0 = time.time()
        lr = lr_cosine_warmup(epoch)
        idx = np.random.choice(len(X_train), size=len(X_train), replace=True, p=p_weights)
        epoch_loss = 0.0
        n_batches  = 0
        coral_outputs_train = []
        coral_outputs_test = []
        for batch_num, start in enumerate(range(0, len(X_train), cfg.batch_size)):
            batch_idx = idx[start:start + cfg.batch_size]
            Xb = X_train[batch_idx]
            yb = y_train[batch_idx].astype(float)
            loss, grad = forward_and_grad(params, Xb, yb)
            # --- MixUp: mix train batch with test features ---
            if MIXUP_ALPHA > 0:
                test_idx = np.random.choice(len(X_test), size=len(Xb), replace=False)
                X_test_batch = X_test[test_idx]
                Xb_mixed, yb_mixed, lam = inter_domain_mixup(Xb, X_test_batch, yb, alpha=MIXUP_ALPHA)
                loss_mixed, _ = forward_and_grad(params, Xb_mixed, yb_mixed)
                loss = loss + loss_mixed  # additional loss from mixed samples
            # --- CORAL: accumulate outputs for domain alignment ---
            if LAMBDA_CORAL > 0:
                # Store training batch outputs
                train_raw = np.array([float(vqc_circuit(x, params)) for x in Xb])
                coral_outputs_train.extend(train_raw)
                # Store test batch outputs
                test_idx_coral = np.random.choice(len(X_test), size=len(Xb), replace=False)
                test_raw = np.array([float(vqc_circuit(x, params)) for x in X_test[test_idx_coral]])
                coral_outputs_test.extend(test_raw)
            t  += 1
            m   = b1 * m + (1 - b1) * grad
            v   = b2 * v + (1 - b2) * grad ** 2
            m_hat = m / (1 - b1 ** t)
            v_hat = v / (1 - b2 ** t)
            params = params - lr * m_hat / (np.sqrt(v_hat) + eps_adam)
            epoch_loss += loss
            n_batches  += 1
        # --- CORAL loss for domain alignment ---
        if LAMBDA_CORAL > 0 and len(coral_outputs_train) > 0:
            # Convert accumulated outputs to torch tensors for coral_loss
            train_t = torch.tensor(np.array(coral_outputs_train).reshape(-1, 1), dtype=torch.float32)
            test_t = torch.tensor(np.array(coral_outputs_test).reshape(-1, 1), dtype=torch.float32)
            coral_val = LAMBDA_CORAL * coral_loss(train_t, test_t)
            epoch_loss += coral_val.item()
            history_vqc.setdefault('coral_loss', []).append(float(coral_val))
        val_raw   = np.array([float(vqc_circuit(x, params)) for x in X_val])
        val_probs = (1 + val_raw) / 2
        val_auc   = roc_auc_score(y_val, val_probs)
        val_loss  = np.mean((val_raw - (2 * y_val - 1)) ** 2)
        history_vqc['train_loss'].append(epoch_loss / max(1, n_batches))
        history_vqc['val_auc'].append(val_auc)
        history_vqc['val_loss'].append(val_loss)
        elapsed = time.time() - t0
        marker = ' *' if val_auc > best_val_auc else ''
        print(f'{epoch+1:5d} | {lr:8.2e} | {epoch_loss/max(1,n_batches):12.4f} | {val_auc:8.4f} | {wait:4d} | {elapsed:6.0f}s{marker}')
        if val_auc > best_val_auc:
            best_val_auc, best_params, wait = val_auc, params.copy(), 0
        else:
            wait += 1
            if wait >= cfg.early_stopping_patience:
                print(f'Early stopping at epoch {epoch+1}. Best val AUC: {best_val_auc:.4f}')
                break
    print('-' * 85)
    print(f'Training complete. Best val AUC: {best_val_auc:.4f} at epoch {len(history_vqc["val_auc"])-wait}')
    np.save(PARAMS_SAVE, best_params)
    with open(f'{cfg.results_dir}/vqc_history.json', 'w') as f:
        json.dump(history_vqc, f, indent=2)
```

### 7.1 QPU Inference (IBM Quantum Hardware)
When `run_mode` is `QPU` or `BOTH`, the trained VQC parameters are deployed to
real IBM Quantum hardware for inference. The circuit is transpiled to the native
gate set of the target backend and executed via Qiskit Runtime SamplerV2.
Zero-Noise Extrapolation (ZNE) via Mitiq is applied for error mitigation.
Results are saved with a `_qpu` suffix for comparison with ideal simulation.

```python
def run_vqc_on_qpu(X, params, backend, shots=1024):
    """Run VQC inference on real IBM Quantum hardware.
    
    Transpiles the PennyLane circuit to the backend's native gate set,
    executes via SamplerV2, and returns probabilities via ⟨Z₀⟩ expectation.
    
    Args:
        X: (n_samples, 64) L2-normalized feature vectors
        params: trained VQC parameters (n_layers, n_qubits, 3)
        backend: IBMQ backend object
        shots: number of measurement shots per circuit
    Returns:
        probs: (n_samples,) numpy array of probabilities
    """
    from qiskit import QuantumCircuit
    from qiskit_ibm_runtime import SamplerV2
    
    probs = []
    n_samples = len(X)
    batch_size = min(50, n_samples)  # submit in batches to avoid queue overload
    
    print(f'Running {n_samples} samples on {backend.name} ({backend.status().pending_jobs} jobs queued)...')
    
    for start in range(0, n_samples, batch_size):
        end = min(start + batch_size, n_samples)
        batch_X = X[start:end]
        
        # Build circuit for each sample
        circuits = []
        for x in batch_X:
            # Use the same circuit structure as vqc_circuit but as a qiskit QuantumCircuit
            # For simplicity, we use pennylane-qiskit to convert
            with qml.queuing.AnnotatedQueue() as q:
                qml.AmplitudeEmbedding(x, wires=range(N_QUBITS), pad_with=0.0)
                for l in range(N_LAYERS):
                    for w in range(N_QUBITS):
                        qml.RY(x[w] * np.pi, wires=w)
                    for w in range(N_QUBITS):
                        qml.Rot(params[l][w][0], params[l][w][1], params[l][w][2], wires=w)
                    for w in range(N_QUBITS):
                        qml.CNOT(wires=[w, (w + 1) % N_QUBITS])
                qml.expval(qml.PauliZ(0))
            
            tape = qml.tape.QuantumScript.from_queue(q)
            circ = qtape_to_circuit(tape)(tape, {})
            
            # Transpile for backend
            circ_tr = transpile(circ, backend=backend, optimization_level=3)
            circ_tr.measure_all()
            circuits.append(circ_tr)
        
        # Execute batch in JOB mode (no Session — Session bills wall-clock time)
        sampler = SamplerV2(mode=backend)
        job = sampler.run(circuits, shots=shots)
        result = job.result()
        
        # Extract ⟨Z₀⟩ from counts
        for circ_idx in range(len(batch_X)):
            counts = result[circ_idx].data.meas.get_counts()
            n_total = sum(counts.values())
            expval = (counts.get('0', 0) - counts.get('1', 0)) / n_total
            prob = (1 + expval) / 2
            probs.append(prob)
        
        print(f'  Batch {start//batch_size + 1}/{(n_samples + batch_size - 1)//batch_size}: '
              f'samples {start}-{end-1} done')
    
    return np.array(probs)
# --- Execute QPU inference if mode is QPU or BOTH ---
if cfg.run_mode in (RunMode.QPU, RunMode.BOTH) and qpu_backend is not None:
    print('\n' + '='*60)
    print('QPU Inference')
    print('='*60)
    
    from qiskit import transpile
    from qiskit_ibm_runtime import QiskitRuntimeService, SamplerV2
    from pennylane_qiskit import qtape_to_circuit
    
    # Run on a subset of the test set (QPU queues are slow)
    n_qpu_samples = min(100, len(X_test))
    np.random.seed(SEED)
    qpu_idx = np.random.choice(len(X_test), size=n_qpu_samples, replace=False)
    X_qpu = X_test[qpu_idx]
    y_qpu = y_test[qpu_idx]
    
    print(f'Evaluating {n_qpu_samples} random test samples on {qpu_backend.name}...')
    
    # Raw hardware probabilities
    qpu_probs = run_vqc_on_qpu(X_qpu, best_params, qpu_backend, shots=cfg.n_qpu_shots)
    np.save(f'{cfg.results_dir}/vqc_qpu_probs.npy', qpu_probs)
    np.save(f'{cfg.results_dir}/vqc_qpu_idx.npy', qpu_idx)
    
    # Ideal simulation on same samples for comparison
    qpu_sim_probs = vqc_predict(X_qpu, best_params)
    np.save(f'{cfg.results_dir}/vqc_qpu_sim_probs.npy', qpu_sim_probs)
    
    # --- ZNE error mitigation ---
    try:
        import mitiq
        from mitiq.zne import execute_with_zne
        from mitiq.zne.inference import RichardsonFactory
        
        # Single job-mode sampler reused across all ZNE folds (avoids opening a
        # new Session per executor call — 100 samples x 3 scale factors = 300 Sessions)
        zne_sampler = SamplerV2(mode=qpu_backend)

        def ibm_executor(circuit):
            """Executor for Mitiq: run circuit on IBM backend, return expval."""
            job = zne_sampler.run([circuit], shots=cfg.n_qpu_shots)
            counts = job.result()[0].data.meas.get_counts()
            n_total = sum(counts.values())
            return (counts.get('0', 0) - counts.get('1', 0)) / n_total
        
        print('Applying ZNE error mitigation (scale factors [1, 2, 3])...')
        qpu_mitigated = []
        for i, x in enumerate(X_qpu):
            # Build circuit
            with qml.queuing.AnnotatedQueue() as q:
                qml.AmplitudeEmbedding(x, wires=range(N_QUBITS), pad_with=0.0)
                for l in range(N_LAYERS):
                    for w in range(N_QUBITS):
                        qml.RY(x[w] * np.pi, wires=w)
                    for w in range(N_QUBITS):
                        qml.Rot(best_params[l][w][0], best_params[l][w][1], best_params[l][w][2], wires=w)
                    for w in range(N_QUBITS):
                        qml.CNOT(wires=[w, (w + 1) % N_QUBITS])
                qml.expval(qml.PauliZ(0))
            tape = qml.tape.QuantumScript.from_queue(q)
            circ = qtape_to_circuit(tape)(tape, {})
            
            mitigated = execute_with_zne(
                circ,
                ibm_executor,
                factory=RichardsonFactory(scale_factors=[1, 2, 3])
            )
            qpu_mitigated.append((1 + mitigated) / 2)
            if (i + 1) % 25 == 0:
                print(f'  ZNE progress: {i+1}/{len(X_qpu)}')
        
        qpu_mitigated = np.array(qpu_mitigated)
        np.save(f'{cfg.results_dir}/vqc_qpu_mitigated_probs.npy', qpu_mitigated)
        print('ZNE error mitigation complete.')
    except ImportError:
        print('Mitiq not installed. Skipping ZNE error mitigation.')
        qpu_mitigated = None
    except Exception as e:
        print(f'ZNE error mitigation failed: {e}')
        qpu_mitigated = None
    
    # --- Comparison table ---
    qpu_auc = roc_auc_score(y_qpu, qpu_probs)
    sim_auc = roc_auc_score(y_qpu, qpu_sim_probs)
    
    print('\n' + '='*60)
    print('QPU vs Simulator Comparison')
    print('='*60)
    print(f'  Simulator (ideal) AUC: {sim_auc:.4f}')
    print(f'  Raw hardware AUC:      {qpu_auc:.4f}')
    if qpu_mitigated is not None:
        mit_auc = roc_auc_score(y_qpu, qpu_mitigated)
        print(f'  ZNE-mitigated AUC:     {mit_auc:.4f}')
    print(f'  Samples evaluated: {n_qpu_samples}')
    print(f'  Backend: {qpu_backend.name}')
    print(f'  Shots per circuit: {cfg.n_qpu_shots}')
    print('='*60)
else:
    print(f'Skipping QPU inference (run_mode={cfg.run_mode.value})')
```

```python
# --- VQC on SVAE features ---
PARAMS_SVAE_SAVE = Path(cfg.results_dir) / 'vqc_svae_best_params.npy'
if PARAMS_SVAE_SAVE.exists():
    print('Loading cached VQC parameters (SVAE)...')
    best_params_svae = np.load(PARAMS_SVAE_SAVE)
else:
    np.random.seed(SEED)
    params_svae = np.random.uniform(0, 2*np.pi, (N_LAYERS, N_QUBITS, 3))
    m_svae  = np.zeros_like(params_svae)
    v_svae  = np.zeros_like(params_svae)
    b1, b2, eps_adam = 0.9, 0.999, 1e-8
    t_svae  = 0
    counts_svae = np.bincount(y_train.astype(int))
    class_weights_svae = 1.0 / np.sqrt(counts_svae)
    sample_weights_svae = class_weights_svae[y_train.astype(int)]
    p_weights_svae = sample_weights_svae / np.sum(sample_weights_svae)
    history_vqc_svae = {'train_loss': [], 'val_auc': [], 'val_loss': [], 'coral_loss': []}
    best_val_auc_svae, best_params_svae, wait_svae = -1.0, params_svae.copy(), 0
    total_batches_svae = int(np.ceil(len(X_train_svae) / cfg.batch_size))
    LAMBDA_CORAL_SVAE = 0.1
    MIXUP_ALPHA_SVAE = 0.2
    print(f'Starting VQC training (SVAE features): {cfg.epochs} epochs, {total_batches_svae} batches/epoch')
    print(f'  LR={cfg.learning_rate}, warmup={cfg.warmup_epochs}, patience={cfg.early_stopping_patience}')
    print(f'  Train: {len(X_train_svae)}, Val: {len(X_val_svae)}')
    print('-' * 85)
    print(f'{"Epoch":>5} | {"LR":>8} | {"Train Loss":>12} | {"Val AUC":>8} | {"Wait":>4} | {"Time":>8}')
    print('-' * 85)
    for epoch in range(cfg.epochs):
        t0 = time.time()
        lr = lr_cosine_warmup(epoch)
        idx = np.random.choice(len(X_train_svae), size=len(X_train_svae), replace=True, p=p_weights_svae)
        epoch_loss = 0.0
        n_batches  = 0
        coral_outputs_train_svae = []
        coral_outputs_test_svae = []
        for batch_num, start in enumerate(range(0, len(X_train_svae), cfg.batch_size)):
            batch_idx = idx[start:start + cfg.batch_size]
            Xb = X_train_svae[batch_idx]
            yb = y_train[batch_idx].astype(float)
            loss, grad = forward_and_grad(params_svae, Xb, yb)
            if MIXUP_ALPHA_SVAE > 0:
                test_idx = np.random.choice(len(X_test_svae), size=len(Xb), replace=False)
                X_test_batch = X_test_svae[test_idx]
                Xb_mixed, yb_mixed, lam = inter_domain_mixup(Xb, X_test_batch, yb, alpha=MIXUP_ALPHA_SVAE)
                loss_mixed, _ = forward_and_grad(params_svae, Xb_mixed, yb_mixed)
                loss = loss + loss_mixed
            if LAMBDA_CORAL_SVAE > 0:
                train_raw = np.array([float(vqc_circuit(x, params_svae)) for x in Xb])
                coral_outputs_train_svae.extend(train_raw)
                test_idx_coral = np.random.choice(len(X_test_svae), size=len(Xb), replace=False)
                test_raw = np.array([float(vqc_circuit(x, params_svae)) for x in X_test_svae[test_idx_coral]])
                coral_outputs_test_svae.extend(test_raw)
            t_svae += 1
            m_svae  = b1 * m_svae + (1 - b1) * grad
            v_svae  = b2 * v_svae + (1 - b2) * grad ** 2
            m_hat = m_svae / (1 - b1 ** t_svae)
            v_hat = v_svae / (1 - b2 ** t_svae)
            params_svae = params_svae - lr * m_hat / (np.sqrt(v_hat) + eps_adam)
            epoch_loss += loss
            n_batches  += 1
        if LAMBDA_CORAL_SVAE > 0 and len(coral_outputs_train_svae) > 0:
            train_t = torch.tensor(np.array(coral_outputs_train_svae).reshape(-1, 1), dtype=torch.float32)
            test_t  = torch.tensor(np.array(coral_outputs_test_svae).reshape(-1, 1), dtype=torch.float32)
            coral_val = LAMBDA_CORAL_SVAE * coral_loss(train_t, test_t)
            epoch_loss += coral_val.item()
            history_vqc_svae.setdefault('coral_loss', []).append(float(coral_val))
        val_raw   = np.array([float(vqc_circuit(x, params_svae)) for x in X_val_svae])
        val_probs = (1 + val_raw) / 2
        val_auc   = roc_auc_score(y_val, val_probs)
        val_loss  = np.mean((val_raw - (2 * y_val - 1)) ** 2)
        history_vqc_svae['train_loss'].append(epoch_loss / max(1, n_batches))
        history_vqc_svae['val_auc'].append(val_auc)
        history_vqc_svae['val_loss'].append(val_loss)
        elapsed = time.time() - t0
        marker = ' *' if val_auc > best_val_auc_svae else ''
        print(f'{epoch+1:5d} | {lr:8.2e} | {epoch_loss/max(1,n_batches):12.4f} | {val_auc:8.4f} | {wait_svae:4d} | {elapsed:6.0f}s{marker}')
        if val_auc > best_val_auc_svae:
            best_val_auc_svae, best_params_svae, wait_svae = val_auc, params_svae.copy(), 0
        else:
            wait_svae += 1
            if wait_svae >= cfg.early_stopping_patience:
                print(f'Early stopping at epoch {epoch+1}. Best val AUC: {best_val_auc_svae:.4f}')
                break
    print('-' * 85)
    print(f'Training complete. Best val AUC: {best_val_auc_svae:.4f} at epoch {len(history_vqc_svae["val_auc"])-wait_svae}')
    np.save(PARAMS_SVAE_SAVE, best_params_svae)
    with open(f'{cfg.results_dir}/vqc_svae_history.json', 'w') as f:
        json.dump(history_vqc_svae, f, indent=2)
# VQC predictions on SVAE features
vqc_svae_val_probs  = vqc_predict(X_val_svae,  best_params_svae)
vqc_svae_test_probs = vqc_predict(X_test_svae, best_params_svae)
np.save(f'{cfg.results_dir}/vqc_svae_val_probs.npy',  vqc_svae_val_probs)
np.save(f'{cfg.results_dir}/vqc_svae_test_probs.npy', vqc_svae_test_probs)
print(f'VQC (SVAE) done. Test probs shape: {vqc_svae_test_probs.shape}')
```

### 8.1 Threshold Selection
Classification thresholds are optimised on the validation set by maximising balanced accuracy, then applied once to the test set to avoid overfitting.

```python
def find_best_threshold(probs, labels, tau_range=np.arange(0.30, 0.80, 0.025)):
    best_tau, best_ba = 0.5, 0.0
    for tau in tau_range:
        preds = (probs > tau).astype(int)
        ba = balanced_accuracy_score(labels, preds)
        if ba > best_ba:
            best_ba, best_tau = ba, tau
    return float(best_tau), float(best_ba)
tau_vqc, ba_vqc_val = find_best_threshold(vqc_val_probs, y_val)
tau_mlp, ba_mlp_val = find_best_threshold(mlp_val_probs, y_val)
```

### 8.2 Main Results
Table 1 presents the full evaluation metrics on the held-out test set.

```python
def compute_all_metrics(probs, labels, tau):
    preds = (probs > tau).astype(int)
    tn, fp, fn, tp = confusion_matrix(labels, preds).ravel()
    return {
        'Accuracy':      round(accuracy_score(labels, preds), 4),
        'Balanced Acc':  round(balanced_accuracy_score(labels, preds), 4),
        'Precision':     round(precision_score(labels, preds), 4),
        'Recall':        round(recall_score(labels, preds), 4),
        'Specificity':   round(tn / (tn + fp), 4),
        'F1':            round(f1_score(labels, preds), 4),
        'AUC-ROC':       round(roc_auc_score(labels, probs), 4),
        'Threshold':     round(tau, 3),
    }
results = {
    'VQC (quantum)':   compute_all_metrics(vqc_test_probs, y_test, tau_vqc),
    'MLP (classical)': compute_all_metrics(mlp_test_probs, y_test, tau_mlp),
}
df_results = pd.DataFrame(results).T
df_results.to_csv(f'{cfg.results_dir}/main_results.csv')
```

```python
N_BOOTSTRAP = 1000
def bootstrap_auc_ci(probs, labels, n=N_BOOTSTRAP, alpha=0.05, seed=SEED):
    rng = np.random.default_rng(seed)
    aucs = []
    for _ in range(n):
        idx = rng.integers(0, len(labels), len(labels))
        if len(np.unique(labels[idx])) < 2:
            continue
        aucs.append(roc_auc_score(labels[idx], probs[idx]))
    lo, hi = np.percentile(aucs, [100 * alpha / 2, 100 * (1 - alpha / 2)])
    return float(lo), float(hi)
```

### 8.9 Cross-Validation Analysis
To ensure robustness of our results beyond a single train/validation/test split, we perform 5-fold stratified cross-validation on the combined training and validation data. This provides mean +/- standard deviation for all metrics, making our findings statistically defensible.

```python
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import (accuracy_score, precision_score, recall_score, 
                             f1_score, balanced_accuracy_score, roc_auc_score)
import json, time
# Combine train + val for CV (stratified on original labels)
X_cv = np.vstack([X_train, X_val])
y_cv = np.concatenate([y_train, y_val])
print(f"Running 5-fold CV on {len(X_cv)} samples...")
n_splits = 5
skf = StratifiedKFold(n_splits=n_splits, shuffle=True, random_state=6)
cv_results = {
    'mlp': {'accuracy': [], 'precision': [], 'recall': [], 'f1': [], 'balanced_acc': [], 'auc_roc': []},
    'vqc': {'accuracy': [], 'precision': [], 'recall': [], 'f1': [], 'balanced_acc': [], 'auc_roc': []}
}
for fold, (train_idx, val_idx) in enumerate(skf.split(X_cv, y_cv)):
    print(f"\n=== Fold {fold+1}/{n_splits} ===")
    X_tr, X_va = X_cv[train_idx], X_cv[val_idx]
    y_tr, y_va = y_cv[train_idx], y_cv[val_idx]
    
    # --- MLP ---
    mlp_fold = nn.Sequential(nn.Linear(64, 32), nn.ReLU(), nn.Dropout(0.3), nn.Linear(32, 1), nn.Sigmoid())
    mlp_fold.to(cfg.device)
    opt_mlp = torch.optim.Adam(mlp_fold.parameters(), lr=1e-3)
    crit_mlp = nn.BCELoss()
    
    for epoch in range(30):
        mlp_fold.train()
        for i in range(0, len(X_tr), 16):
            batch_x = torch.tensor(X_tr[i:i+16], dtype=torch.float32).to(cfg.device)
            batch_y = torch.tensor(y_tr[i:i+16], dtype=torch.float32).to(cfg.device)
            opt_mlp.zero_grad()
            out = mlp_fold(batch_x).squeeze()
            loss = crit_mlp(out, batch_y)
            loss.backward()
            opt_mlp.step()
    
    mlp_fold.eval()
    with torch.no_grad():
        mlp_probs = mlp_fold(torch.tensor(X_va, dtype=torch.float32).to(cfg.device)).cpu().numpy().squeeze()
    mlp_preds = (mlp_probs > 0.5).astype(int)
    
    cv_results['mlp']['accuracy'].append(accuracy_score(y_va, mlp_preds))
    cv_results['mlp']['precision'].append(precision_score(y_va, mlp_preds, zero_division=0))
    cv_results['mlp']['recall'].append(recall_score(y_va, mlp_preds, zero_division=0))
    cv_results['mlp']['f1'].append(f1_score(y_va, mlp_preds, zero_division=0))
    cv_results['mlp']['balanced_acc'].append(balanced_accuracy_score(y_va, mlp_preds))
    cv_results['mlp']['auc_roc'].append(roc_auc_score(y_va, mlp_probs))
    
    # --- VQC ---
    np.random.seed(42 + fold)
    params_fold = np.random.uniform(-0.1, 0.1, (3, 6, 3))
    
    best_vqc_auc, best_params_fold = 0, params_fold.copy()
    wait_fold = 0
    
    for epoch in range(15):
        indices = np.random.permutation(len(X_tr))
        for i in range(0, min(256, len(X_tr)), 8):
            batch_idx = indices[i:i+8]
            X_batch = X_tr[batch_idx]
            y_batch = y_tr[batch_idx]
            
            grad = np.zeros_like(params_fold)
            for j, x in enumerate(X_batch):
                for layer in range(3):
                    for qubit in range(6):
                        for k in range(3):
                            params_test = params_fold.copy()
                            params_test[layer, qubit, k] += np.pi/2
                            plus = float(vqc_circuit(x, params_test))
                            params_test[layer, qubit, k] -= np.pi/2
                            minus = float(vqc_circuit(x, params_test))
                            grad[layer, qubit, k] += (plus - minus) * (2 * y_batch[j] - 1) / 8
            params_fold -= 1e-3 * grad
        
        vqc_val_probs = np.array([float(vqc_circuit(x, params_fold)) for x in X_va])
        vqc_val_auc = roc_auc_score(y_va, (1 + vqc_val_probs) / 2)
        
        if vqc_val_auc > best_vqc_auc:
            best_vqc_auc, best_params_fold = vqc_val_auc, params_fold.copy()
            wait_fold = 0
        else:
            wait_fold += 1
            if wait_fold >= 3:
                break
    
    vqc_probs = np.array([float(vqc_circuit(x, best_params_fold)) for x in X_va])
    vqc_probs = (1 + vqc_probs) / 2
    vqc_preds = (vqc_probs > 0.5).astype(int)
    
    cv_results['vqc']['accuracy'].append(accuracy_score(y_va, vqc_preds))
    cv_results['vqc']['precision'].append(precision_score(y_va, vqc_preds, zero_division=0))
    cv_results['vqc']['recall'].append(recall_score(y_va, vqc_preds, zero_division=0))
    cv_results['vqc']['f1'].append(f1_score(y_va, vqc_preds, zero_division=0))
    cv_results['vqc']['balanced_acc'].append(balanced_accuracy_score(y_va, vqc_preds))
    cv_results['vqc']['auc_roc'].append(roc_auc_score(y_va, vqc_probs))
    
    print(f"MLP: Acc={cv_results['mlp']['accuracy'][-1]:.4f}, AUC={cv_results['mlp']['auc_roc'][-1]:.4f}")
    print(f"VQC: Acc={cv_results['vqc']['accuracy'][-1]:.4f}, AUC={cv_results['vqc']['auc_roc'][-1]:.4f}")
print("\n" + "="*60)
print("5-FOLD CROSS-VALIDATION RESULTS (mean +/- std)")
print("="*60)
for model in ['mlp', 'vqc']:
    print(f"\n{model.upper()}:")
    for metric in ['accuracy', 'precision', 'recall', 'f1', 'balanced_acc', 'auc_roc']:
        vals = cv_results[model][metric]
        print(f"  {metric:15s}: {np.mean(vals):.4f} +/- {np.std(vals):.4f}")
with open(f'{cfg.results_dir}/cv_results.json', 'w') as f:
    json.dump(cv_results, f, indent=2)
print(f"\nCV results saved to {cfg.results_dir}/cv_results.json")
```

### 9.1 Summary of Findings
This work demonstrates that a hybrid quantum–classical pipeline can perform binary pneumonia detection from chest X-ray images with performance statistically indistinguishable from a classical MLP baseline, while using substantially fewer trainable parameters. The key findings are:
1. **Parameter efficiency.** The VQC achieves competitive performance with only 108 trainable parameters compared to the MLP's 2,145—a $20\times$ reduction. This supports the hypothesis that quantum circuits can serve as compact function approximators.
2. **Ansatz selection matters.** The expressibility and entanglement analysis (Section 5) provides a principled basis for selecting circuit depth, rather than relying on grid search. The chosen configuration ($L = 3$) represents a balance between expressibility and trainability.
3. **No quantum advantage observed.** While the VQC matches the classical baseline, it does not exceed it. This is consistent with the current understanding of NISQ-era QML: quantum models are competitive on small-scale problems but have not yet demonstrated clear advantages over well-tuned classical methods.
### 9.2 Limitations
Several limitations should be noted. First, the dataset, while standard in the literature, is relatively small and may not generalise to diverse clinical populations. Second, all quantum simulations are performed on classical hardware; evaluation on actual quantum hardware would introduce noise effects not captured in this study. Third, the PCA dimensionality reduction step, while necessary for amplitude encoding, discards information that may be relevant for classification.
### 9.3 Future Work
Promising directions include: (i) evaluation on real quantum hardware with error mitigation, (ii) exploration of alternative encoding schemes (e.g., angle encoding, basis encoding) that do not require PCA, (iii) scaling to larger datasets and multi-class classification tasks, and (iv) investigating whether quantum advantage emerges at larger problem sizes where classical simulation becomes intractable.

## 10. Conclusion
We presented a hybrid quantum–classical pipeline for pneumonia detection from chest radiographs, combining a frozen ConvNeXt-Tiny feature extractor with a data re-uploading variational quantum circuit. Through formal expressibility and entanglement analysis, we justified our architectural choices and demonstrated that the quantum classifier achieves statistically equivalent performance to a classical MLP baseline with $20\times$ fewer trainable parameters. While no quantum advantage was observed, this work establishes a rigorous evaluation framework for comparing quantum and classical models on real-world medical imaging tasks, and provides a reproducible baseline for future studies as quantum hardware continues to mature.

## 8. Analysis: DANN Domain Invariance Visualization
__Analysis__: Did the DANN successfully learn domain-invariant features?
__Method__: Extract DANN features → t-SNE visualization → Train domain classifier
__Interpretation__:
- Domain classifier accuracy ~50% = DANN SUCCESS (features intermix)
- Domain classifier accuracy >55% = Domain information remains

```python
# Analysis 1: DANN Domain Invariance Visualization
from sklearn.manifold import TSNE
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
print('='*60)
print('Analysis 1: DANN Domain Invariance')
print('='*60)
# Load DANN features from cache
FEAT_DIR = Path('artifacts/features')
cache_train = FEAT_DIR / 'convnext_tiny_train.npy'
cache_test = FEAT_DIR / 'convnext_tiny_test.npy'
if cache_train.exists() and cache_test.exists():
    print('Loading cached DANN features...')
    train_feat = np.load(cache_train)
    test_feat = np.load(cache_test)
else:
    print('ERROR: DANN features not found. Run DANN training first.')
    raise FileNotFoundError('Run DANN training in Colab first.')
print(f'Train features: {train_feat.shape}, Test features: {test_feat.shape}')
# t-SNE visualization
def plot_domain_invariance(train_feat, test_feat, save_path):
    all_feat = np.vstack([train_feat, test_feat])
    domains = np.array([0]*len(train_feat) + [1]*len(test_feat))
    tsne = TSNE(n_components=2, random_state=SEED, perplexity=50)
    feat_2d = tsne.fit_transform(all_feat)
    plt.figure(figsize=(10, 8))
    colors = ['#1f77b4', '#d62728']
    labels = ['Train (Source, 74.2% pn.)', 'Test (Target, 62.5% pn.)']
    for d in [0, 1]:
        mask = domains == d
        plt.scatter(feat_2d[mask, 0], feat_2d[mask, 1], c=colors[d], label=labels[d], alpha=0.5, s=20)
    plt.title('DANN Feature Space: Domain Invariance Test', fontsize=14)
    plt.xlabel('t-SNE Component 1')
    plt.ylabel('t-SNE Component 2')
    plt.legend()
    plt.savefig(save_path, bbox_inches='tight')
    plt.close()
    return feat_2d
# Domain classifier
def test_domain_classifier(train_feat, test_feat):
    X = np.vstack([train_feat, test_feat])
    y = np.array([0]*len(train_feat) + [1]*len(test_feat))
    clf = LogisticRegression(max_iter=1000, random_state=SEED)
    clf.fit(X, y)
    preds = clf.predict(X)
    acc = accuracy_score(y, preds)
    return acc
# Run analysis
FIGS_DIR = cfg.figures_dir
tsne_feat = plot_domain_invariance(train_feat, test_feat, save_path=FIGS_DIR / 'dann_domain_tsne.png')
domain_acc = test_domain_classifier(train_feat, test_feat)
print()
print('='*60)
RESULTS:
print('='*60)
print(f'Domain classifier accuracy: {domain_acc:.1%}')
if domain_acc < 0.55:
    print('✓ DANN SUCCESS: Features are domain-invariant (~50% = random)')
else:
    print('✗ Domain shift partially remaining (>55% = can distinguish)')
print('='*60)
```
