# 03e — Grad-CAM Interpretability Analysis

**Purpose:** Visualize model attention using Grad-CAM to understand what features the model uses for predictions.

| Input | Output |
|---|---|
| `results/figures/gradcam/*.png` | Attention visualizations |

**Runtime:** <1 minute (CPU only)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
from PIL import Image
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
plt.rcParams['figure.dpi'] = 120

RESULTS_DIR = Path('results')
FIG_DIR = RESULTS_DIR / 'figures' / 'gradcam'
for d in [FIG_DIR]:
    d.mkdir(parents=True, exist_ok=True)

CLASS_NAMES = ['Normal', 'Pneumonia']
CLASS_COLORS = ['#1B4F8A', '#C0392B']
```

## 1. Load Grad-CAM Visualizations

```python
# Look for saved Grad-CAM images
gradcam_files = list(FIG_DIR.glob('*.png')) + list(FIG_DIR.glob('*.jpg'))
print(f'Found {len(gradcam_files)} Grad-CAM files')

if not gradcam_files:
    print('\nNo Grad-CAM images found.')
    print('These should be generated by main notebook (pneumonia_hybrid_qml.ipynb)')
    print('Expected location: results/figures/gradcam/*.png')
```

## 2. Display Template (when available)

```python
def display_gradcams(gradcam_dir, save_dir):
    """Display and analyze saved Grad-CAM images."""
    files = sorted(gradcam_dir.glob('*.png'))
    
    if not files:
        return
    
    # Sample grid
    n = min(8, len(files))
    fig, axes = plt.subplots(2, 4, figsize=(14, 7))
    axes = axes.flatten()
    
    for i, f in enumerate(files[:n]):
        img = Image.open(f)
        axes[i].imshow(img)
        axes[i].axis('off')
        axes[i].set_title(f.name[:30], fontsize=9)
    
    for i in range(n, len(axes)):
        axes[i].axis('off')
    
    plt.tight_layout()
    plt.savefig(save_dir / 'gradcam_overview.pdf', bbox_inches='tight')
    plt.show()
    print(f'saved → {save_dir}/gradcam_overview.pdf')

    return len(files)

if gradcam_files:
    n_images = display_gradcams(FIG_DIR, FIG_DIR)
    print(f'\nDisplayed {n_images} Grad-CAM visualizations')
```

## 3. Attention Pattern Analysis Template

```python
def analyze_attention_patterns(gradcam_dir, save_dir):
    """Template for analyzing attention patterns when images available."""
    files = list(gradcam_dir.glob('*.png'))
    
    if len(files) < 2:
        return None
    
    # Analyze file naming patterns
    # Expected: class_predicted_*.png
    normal_files = [f for f in files if 'normal' in f.name.lower()]
    pneumo_files = [f for f in files if 'pneumonia' in f.name.lower() or 'pneumo' in f.name.lower()]
    
    print('=== Grad-CAM Analysis ===')
    print(f'  Normal images: {len(normal_files)}')
    print(f'  Pneumonia images: {len(pneumo_files)}')
    
    # Save summary
    import json
    summary = {
        'total_images': len(files),
        'normal_count': len(normal_files),
        'pneumonia_count': len(pneumo_files)
    }
    
    with open(save_dir / 'gradcam_summary.json', 'w') as f:
        json.dump(summary, f, indent=2)
    
    print(f'saved → {save_dir}/gradcam_summary.json')
    return summary

if gradcam_files:
    analyze_attention_patterns(FIG_DIR, FIG_DIR)
```
